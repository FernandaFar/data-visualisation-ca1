## Selecting National Employment Data

Since the dataset has information from different areas of the United States, I decided to focus on national cross-industry data. This allows me to compare employment and wage information for the selected occupations at a national level.

```python
national_data = dataset.loc[
    (dataset['AREA_TITLE'] == 'U.S.') &
    (dataset['NAICS_TITLE'] == 'Cross-industry') &
    (dataset['OCC_TITLE'].isin([
        'Data Scientists',
        'Software Developers',
        'Computer and Information Research Scientists'
    ]))
]
```

```python
national_data[['OCC_CODE', 'OCC_TITLE', 'TOT_EMP', 'H_MEDIAN']]
```

The dataset contains both broader occupational categories and more specific occupation codes. Some employment and wage values appear more than once because of this classification structure. To avoid double-counting, I will use the more specific occupation codes for the comparison.

```python
national_data = national_data.loc[
    (national_data['OCC_CODE'] == '15-1221') |
    (national_data['OCC_CODE'] == '15-1252') |
    (national_data['OCC_CODE'] == '15-2051')
]

national_data[['OCC_CODE', 'OCC_TITLE', 'TOT_EMP', 'H_MEDIAN']]
```
